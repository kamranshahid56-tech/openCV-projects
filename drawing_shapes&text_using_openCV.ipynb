```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
blank = np.zeros((500,500,3), dtype = "uint8")
```

```python
blank
```

```python
plt.imshow(blank)
```

```python
blank[:] = 255,0,0
```

```python
plt.imshow(blank)
```

```python
blank[150:300, 320:400] = 0,255,0
```

```python
plt.imshow(blank)
```

```python
rect = cv2.rectangle(blank,(100,200),(250,250),(200,100,245), thickness = 2)
```

```python
plt.imshow(rect)
```

```python
blank1 = np.zeros((500,500,3),dtype = "uint8")
```

```python
plt.imshow(blank1)
```

```python
circle = cv2.circle(blank1, (blank1.shape[1]//2,blank.shape[0]//2), 40, (0,0,255), thickness = 5) 
```

```python
plt.imshow(circle)
```

```python
line = cv2.line(blank1, (500,500), (blank1.shape[1]//2,blank1.shape[0]//2), (255,255,255), thickness = 10)
```

```python
plt.imshow(line)
```

```python
text = np.zeros((500,500,3),dtype = "uint8")
```

```python
texted = cv2.putText(text, "Hello World", (100,250),cv2.FONT_HERSHEY_TRIPLEX,1.0,(250,250,250),thickness = 2)
```

```python
plt.imshow(texted)
```
