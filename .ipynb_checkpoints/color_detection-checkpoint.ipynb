```python
import cv2
import numpy as np
```

```python
cam = cv2.VideoCapture(0)
```

```python
while(1):
    res,frame = cam.read()
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    red_L = np.array([135,85,115], np.uint8)
    red_U = np.array([185,250,250], np.uint8)
    redM = cv2.inRange(hsv, red_L, red_U)

    green_L = np.array([30,55,75], np.uint8)
    green_U = np.array([105,250,250], np.uint8)
    greenM = cv2.inRange(hsv, green_L, green_U)

    
    kernel = np.ones((5,5), "uint8")
    
    redM = cv2.dilate(redM, kernel)
    red = cv2.bitwise_and(frame,frame, mask = redM)

    greenM = cv2.dilate(greenM, kernel)
    green = cv2.bitwise_and(frame,frame, mask = greenM)

    
    cont,heir = cv2.findContours(redM, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
    for pic,cont in enumerate(cont):
        area = cv2.contourArea(cont)
        if (area>300):
            x,y,w,h = cv2.boundingRect(cont)
            frame = cv2.rectangle(frame,(x,y),(x+w,y+h),(0,0,255),2)
            cv2.putText(frame, "Red Color", (x,y),cv2.FONT_HERSHEY_TRIPLEX,1.0,(0,0,255))

    cont1,heir1 = cv2.findContours(greenM, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
    for pic,cont1 in enumerate(cont1):
        area1 = cv2.contourArea(cont1)
        if (area1>300):
            x,y,w,h = cv2.boundingRect(cont1)
            frame = cv2.rectangle(frame,(x,y),(x+w,y+h),(0,255,0),2)
            cv2.putText(frame, "Green Color", (x,y),cv2.FONT_HERSHEY_TRIPLEX,1.0,(0,255,0))

            
    cv2.imshow("Red Color", frame)
    if cv2.waitKey(1) & 0xff==ord('q'):
        cam.release()
        cv2.destroyAllWindows()
        break
            
    
```
