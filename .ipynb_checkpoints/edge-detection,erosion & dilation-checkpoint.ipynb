```python
import cv2
```

```python
img = cv2.imread('quest.jpeg')
img = cv2.resize(img,(720,400))
```

```python
cv2.imshow('Image', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
cv2.imshow('Image', gray_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
canny_img = cv2.Canny(gray_img, 150,150)
```

```python
cv2.imshow('Image', canny_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
video = cv2.VideoCapture(0)
```

```python
while True:
    res,frame = video.read()
    frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    if not res:
        break
    cv2.imshow("Video", frame)

    if cv2.waitKey(1) & 0xff==ord('q') or cv2.getWindowProperty("Video", cv2.WND_PROP_VISIBLE) < 1:
        break

video.release()
cv2.destroyAllWindows()
```

```python
import numpy as np
kernel = np.ones((2,2), np.uint8)
```

```python
erode_img = cv2.erode(canny_img, kernel, iterations = 1)
```

```python
cv2.imshow('Image', erode_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
dilate_img = cv2.dilate(canny_img, kernel, iterations = 1)
```

```python
cv2.imshow('Image', dilate_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
