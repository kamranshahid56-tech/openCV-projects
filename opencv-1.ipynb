```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
```

```python
cv2.imshow("HSV", hsv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
video = cv2.VideoCapture(0)
```

```python
while True:
    res, frame = video.read()
    
    # Check if the video has ended or the camera is disconnected
    if not res:
        print("Empty frame or video ended.")
        break

    cv2.imshow("Video", frame)

    # getWindowProperty checks if the window actually exists
    # If the user clicks 'X', the property returns -1
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
